#### Random Forest Regression

```python
# Importing the necessary Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataset

dataset = pd.read_csv("2.2 Gaming_data.csv.csv")

# Split the dataset into target and variables
X = dataset.iloc[:,0:1].values
y = dataset.iloc[:,1:2].values
```

```python
# Fitting the Random Forest Regressor to the dataset
from sklearn.ensemble import RandomForestRegressor 

# create the Forest Object
regressor = RandomForestRegressor(n_estimators=10,random_state=0)
regressor.fit(X,y.ravel())
```

```python
# Visualize the Results

plt.scatter(X,y)
plt.plot(X, regressor.predict(X), color='red')
plt.title("Gaming Data (Random Forest)")
plt.xlabel("Gaming Data")
plt.ylabel("Points")
plt.show()
```

```python
# Visualize the results in higher resolution

X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y)
plt.plot(X_grid, regressor.predict(X_grid), color='red')
plt.title('Gaming data (Random Forest)')
plt.xlabel('Gaming steps')
plt.ylabel('Points')
plt.show()
```

```python
# Make prediction using the random forest with 10 trees
y_pred = regressor.predict([[8.5]])
print(y_pred)
```

```python
# Let's retrain a random forest model with 100 trees and predict
# Fitting the Random Forest Regressor to the dataset
from sklearn.ensemble import RandomForestRegressor 

# create the Forest Object
regressor = RandomForestRegressor(n_estimators=100,random_state=0)
regressor.fit(X,y.ravel())

# Visualize the results in higher resolution

X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y)
plt.plot(X_grid, regressor.predict(X_grid), color='red')
plt.title('Gaming data (Random Forest)')
plt.xlabel('Gaming steps')
plt.ylabel('Points')
plt.show()

# Make prediction using the random forest with 100 trees
y_pred = regressor.predict([[8.5]])
print(y_pred)
```

```python
# Let's finally retrain a random forest model with 300 trees and predict
# Fitting the Random Forest Regressor to the dataset
from sklearn.ensemble import RandomForestRegressor 

# create the Forest Object
regressor = RandomForestRegressor(n_estimators=300,random_state=0)
regressor.fit(X,y.ravel())

# Visualize the results in higher resolution

X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y)
plt.plot(X_grid, regressor.predict(X_grid), color='red')
plt.title('Gaming data (Random Forest)')
plt.xlabel('Gaming steps')
plt.ylabel('Points')
plt.show()

# Make prediction using the random forest with 300 trees
y_pred = regressor.predict([[8.5]])
print(y_pred)
```
